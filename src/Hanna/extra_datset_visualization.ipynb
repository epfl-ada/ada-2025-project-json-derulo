```python
#dataset_path = file_path = pd.read_txt(r"C:\Users\hanna\ADA\2025\newyorker_caption_contest\data\510.csv")
dataset_path = r"C:\Users\hanna\ADA\2025\additional_dataset\GDELT.MASTERREDUCEDV2.txt"
from itertools import islice

with open(dataset_path, "r") as f:
    for line in f.readlines()[:5]:
        print(line.strip())
```

```python
# Adjust separator if needed (tab or space). Try '\t' first.
chunksize = 100000  # adjust depending on your memory
found_rows = []

for chunk in pd.read_csv(dataset_path, sep='\t', chunksize=chunksize, dtype=str):
    # Filter rows where Date starts with "2010" and Source or Target contains "USA"
    filtered = chunk[
        chunk['Date'].str.startswith('2010', na=False) &
        (chunk['Source'].str.contains('USA', na=False) | chunk['Target'].str.contains('USA', na=False))
    ]
    
    if not filtered.empty:
        found_rows.append(filtered)
    
    # Stop early if we already have at least 5 rows
    if sum(len(f) for f in found_rows) >= 5:
        break

# Concatenate and show only the first 5
if found_rows:
    result = pd.concat(found_rows).head(5)
    print(result)
else:
    print("No matching rows found.")
```
